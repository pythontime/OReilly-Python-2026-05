# Jupyter and writing Python

If you want to program in Python, then you need:

1. Python installed somewhere
2. An editor that lets you write in Python

The problem is that installing Python can be a pain. And installing an editor can also be one.

I use an editor known as Jupyter, which is a "notebook." It's not a file that I type and then run, but it's more of stream-of-consciousness programming. I type into a cell, and that cell can contain either text (Markdown) or Python code. Then I can "execute" it -- if it's Python, then it runs, and if it's Markdown, then it is displayed nicely.

If you want to run Python code, you can (a) install Python, (b) install Jupyter or an editor such as PyCharm or VSCode, and (c) write in it.

I have some videos on the O'Reilly platform of how to install Python and VSCode.

A much, much easier way to do things is to use Jupyter Lite:

- It only exists in your browser
- It doesn't require installing Python, because it uses a version of Python that is *in* your browser
- It also provides the entire notebook system

Here's what to do:

- Go to https://jupyter.org/try-jupyter/lab/
- Click on Python/Pyodide
- This gives you a notebook

# Quick intro to Jupyter

- We use notebooks
- Every notebook is a separate, continously running Python program
- The notebook is broken into cells

What can we put in cells?

- Python code
- Markdown (which is an easy-to-write version of HTML)

# Writing in Jupyter

When you type into a cell, you are in one of two modes:

- In *edit* mode, anything you type is put into the cell. Like right now, as I'm typing. (This is the normal mode.) To get into edit mode, just click in a cell or press ENTER.
- In *command* mode, anything you type is interpreted as a command for Jupyter. These are typically one-character commands. What you type changes Jupyter's behavior, but doesn't add any text to the notebook. To enter command mode, click to the left of a cell or press ESC.

A few more things:
- To execute a cell, press shift+ENTER at the same time.

Some command-mode commands:

- `x` -- cut a cell
- `c` -- copy a cell
- `v` -- paste a cell
- `a` -- create a new cell *above* the current one
- `b` -- create a new cell *below* the current one
- `m` -- change the cell to Markdown documentation
- `y` -- change the cell to Python code
