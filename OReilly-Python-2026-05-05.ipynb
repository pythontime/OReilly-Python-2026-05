# Writing in Jupyter

When you type into a cell, you are in one of two modes:

- In *edit* mode, anything you type is put into the cell. Like right now, as I'm typing. (This is the normal mode.) To get into edit mode, just click in a cell or press ENTER.
- In *command* mode, anything you type is interpreted as a command for Jupyter. These are typically one-character commands. What you type changes Jupyter's behavior, but doesn't add any text to the notebook. To enter command mode, click to the left of a cell or press ESC.

A few more things:
- To execute a cell, press shift+ENTER at the same time.

Some command-mode commands:

- `x` -- cut a cell
- `c` -- copy a cell
- `v` -- paste a cell
- `a` -- create a new cell *above* the current one
- `b` -- create a new cell *below* the current one
- `m` -- change the cell to Markdown documentation
- `y` -- change the cell to Python code

# What is a programming language? What is Python?

When computers were first invented, they were single-purpose machines. If you wanted to change how a computer calculated things or reported its results, you needed to rewire it, and maybe change/add/remove components. 

Very quickly, people realized that having a general-purpose computer that could be reprogrammed as needed was a much better solution. That's where programming languages came in.

Originally, all programming was done with 1s and 0s ("binary code"), which is what the computer uses for its instructions and data. But that's a pain to work with. A programming language lets us write instructions in a higher-level form that closer to human language, and then translates that into 1s and 0s.

There are many thousands of programming languages. Each has different advantages and disadvantages:

- C is super fast to execute, but hard for people to write, and to get right
- C++ is also fast, and gives you an opportunity to write more complex data structures using "object-oriented programming," but is even harder than C.
- Java and C# were invented to give the power of C++ without the overhead
- Python was deliberately invented to give the power of other languages without the steep learning curve or difficulty.

# Variables and assignment

It's annoying and boring to use the same values repeatedly. What we want to do is allow for different values in our program. For that, we need *variables*. 

If functions are verbs and data is nouns, then variables are pronouns, referring to a noun (a value).

We can assign a variable to a value simply by using the `=`, the "assignment operator":

- A variable is on the left
- `=` is in the middle
- The value to which we want to assign the variable is on the right

Once we have assigned, the variable will continue to refer to that value until you change it or the program exits.

```python
x = 10    # here, I assign the variable x the numeric value 10

print(x + 5)  # here, we display the result of adding x (10) to 5
```

```python
x = 10
y = 5

print(x + y)
```

# Assignment with `=`

This operator, `=`, is *not* the same as the `=` you use in mathematics! (It's the same symbol, but used in a very different way.) 

When you assign, the value on the right is assigned to the variable on the left.

Where do we declare `x` to be a variable, or what type of values it can take? Answer: The first time you assign to a variable, it is created. No declaration is necessary (or really possible). 

The second time you assign to a variable, it doesn't need to be created, and the value is just assigned to it.
