# Variables and assignment

It's annoying and boring to use the same values repeatedly. What we want to do is allow for different values in our program. For that, we need *variables*. 

If functions are verbs and data is nouns, then variables are pronouns, referring to a noun (a value).

We can assign a variable to a value simply by using the `=`, the "assignment operator":

- A variable is on the left
- `=` is in the middle
- The value to which we want to assign the variable is on the right

Once we have assigned, the variable will continue to refer to that value until you change it or the program exits.

````python
x = 10    # here, I assign the variable x the numeric value 10

print(x + 5)  # here, we display the result of adding x (10) to 5
````

````python
x = 10
y = 5

print(x + y)
````

# Assignment with `=`

This operator, `=`, is *not* the same as the `=` you use in mathematics! (It's the same symbol, but used in a very different way.) 

When you assign, the value on the right is assigned to the variable on the left.

Where do we declare `x` to be a variable, or what type of values it can take? Answer: The first time you assign to a variable, it is created. No declaration is necessary (or really possible). 

The second time you assign to a variable, it doesn't need to be created, and the value is just assigned to it.



# Variable names

Python's rules for variable names:

- You can use letters (capital and lowercase), digits, and `_`.

- Variable names can start with either `_` or a letter (but not a digit)
- In general, you don't want to start or end variable names with `_`, because that can be an indication that it's an internal, or special, variable name.

You should always use long, clear variable names.

# Conditions 

Almost all of programming is about making decisions. And we now have the tools we need to start making those decisions.

We will use `if` to decide whether some code should run. Until now, there was no way to say that a portion of a program should not run. It all ran, from start to finish.

You can now say `if` (some condition is `True`), and then that code will only execute in such a case.

The idea is that `if` looks to its right. If it sees a `True` value, then its "block" of code will run. If it sees a `False` value, then execution continues with the `else` block that is associated with it.

````python
name = input('Enter your name: ')

if name == 'Reuven':
    print('Hello, boss!')
    print('It is great to see you again!')
else:
    print('Hello, ' + name + '.')
````

# Reviewing this code

1. We assign to `name` based on the user's input.
2. `if` looks to its right, and checks if there is a `True` value.
3. It'll see a `True` value if `name == 'Reuven'`.
4. If the `if` sees `True`, then it executes its block.
5. Notice that at the end of the `if` line, there is a `:`. That is *mandatory*. It comes before a block of code.
6. A block of code is indented, typically with 4 spaces. Python doesn't use `{}` or `begin/end` to mark the start and end of a block. Rather, it's `:` followed by indentation. That indentation can be any number of spaces or tabs, *but* we normally use 4 spaces *and* your editing tool will probably indent for you automatically.
7. When you're done with an indented block, just press backspace/delete. Keep coding from there. An indented block can contain any number of lines.
8. A block can contain *ANY* Python code whatsoever, including `print`, `input`, or even more `if` statements.
9. An `else` block is optional, but common. It does *not* take any condition. It only runs if the `if` was `False`. Its block then runs.

If you have an `if`/`else` statement, then one -- and only one -- of these will run. It cannot be that neither runs. It cannot be that both run. One of them will run.

# What if I have more than two possibilities?

If I have three or more possibilities, then I can use `elif`. This goes between `if` and `else` (although `else` is still optional). If `if`'s condition is `False`, then we try the first `elif`. If that is `False`, we keep going until either one is `True` or we get to the `else`. If there is no `else`, then maybe nothing will happen.

````python
name = input('Enter your name: ')

if name == 'Reuven':
    print('Hello, boss!')
    print('It is great to see you again!')

elif name == 'something else':
    print('That is a weird name, but ... sure!')

else:
    print('Hello, ' + name + '.')
````

You can have as many `elif` clauses as you want. Again: The first `if` or `elif` whose condition is `True` fires, and **NONE** of the others do. `else` runs if all of the `if` and `elif` clauses are `False`.

# Complex conditions

It's nice that you can have `if`/`elif`/`else`. But sometimes you don't want to check just one thing. Sometimes, you want to check multiple things.

For that, you have `and` and `or` in Python. (In some languages, you use special symbols. Not in Python!)



````python
# and returns True if the expressions on both its left and right are True

x = 10
y = 20

#    True  and  True --> True
if x == 10 and y == 20:
    print('Yes! Both are what you want!')
````

````python
#   False    and  True --> False
if x == 100 and y == 20:
    print('Yes! Both are what you want!')   # there is no else, so nothing will print
````

````python
# sometimes, though, you want to check if one of several options is True.
# If one or more is True, that's good enough -- it doesn't have to be all of them.

x = 10
y = 20

#    True  or   True --> True
if x == 10 or y == 20:
    print('Yes! At least one is what you want')
````

````python
x = 10
y = 20

#    False  or   True --> True
if x == 100 or y == 20:
    print('Yes! At least one is what you want')
````

````python

x = 10
y = 20

#    True  or   False    --> True
if x == 10 or y == 200:
    print('Yes! At least one is what you want')
````

````python
# briefly, we also have a "not" operator in Python
# it goes before an expression
# it reverses, or flips, the logic of that expression
# True becomes False, and False becomes True

x = 10

if not x == 20:
    print('It is not 20!')
````

````python
# don't do that! Just use != instead!

if x != 20:
    print('It is not 20!')
````

# f-strings

So far, we have found that strings are easy to create... unless they include variables. And then things get messy:

f-strings -- short for *format* strings -- are a fantastic invention in Python from about 10 years ago. The idea is that you put an `f` before the opening quotes, and then inside of your string, you can put `{}`. Inside of the `{}`, you can put any Python expression, including a variable. It is transformed into a string, if it isn't one already.

````python
x = 10

print(f'The value of x is {x}.')  
````

````python
x = 10
y = 20

print(f'{x} + {y} = {x+y}')
````

````python
# you could do this:

first = input('Enter first word: ')
second = input('Enter second word: ')

if first < second:   
    print(f'{first} comes before {second}.')
elif second < first:
    print(f'{second} comes before {first}.')
elif first == second:
    print(f'{first} and {second} are the same.')
else:
    print('You should never see this!')
````

2. Ask the user (just once) to guess `secret`.
3. Give one of three responses:
    - Too low
    - Too high
    - You got it!

This is a very dumb game. The user only gets one guess.

````python
secret = 72   # this is an integer, because it's only digits, and no quotes!

guess = input('Guess: ')

if guess == secret:
    print('You got it!')
elif guess < secret:
    print('Too low!')
else:
    print('Too high!')
````

````python
# for this to work, I need to get an integer based on the user's input

secret = 72   # this is an integer, because it's only digits, and no quotes!

guess = input('Guess: ')
guess = int(guess)  # I could say int(input('Guess: ')), but I prefer this for several reasons

if guess == secret:
    print('You got it!')
elif guess < secret:
    print('Too low!')
else:
    print('Too high!')
````

# Strings

Programmers talk about "strings," and they mean "text." In Python, a string can contain zero characters (the "empty string"), or one character, or ... any number you want. The only limit on the size of a string is the memory in your computer.

There is no "character" type in Python, but we do talk about characters.

Python strings, use Unicode, meaning that they can include any character from any language, as well as many symbols, flags, etc., including emojis.

We define a string with either `''` or `""`. If you want to include quotes in your string, you have to use the other type -- otherwise, Python will get confused over which quotes are inside of the string, and which are forming its borders. You can, if you want, use `\` before a quote to "escape" it, and make it literal. 

Meaning, you could say:

````python
s = 'he's very nice'
````

````python
s = 'he\'s very nice'   # \' is a literal quote in a string
````

````python
s
````

````python
s = 'abcdefghijklmnopqrstuvwxyz'

# how many characters are in a string? Use the len() function to find out
len(s)
````

````python
# what if I want one particular character?
# I can retrieve it with [], putting the index of the character inside
# note that the index starts at 0 

s[0]  # this returns the first character in s
````

````python
s[1]
````

````python
s[25]   # this is 'z', the final character
````

````python
s[26]   
````

We see that `len(s)` returns 1 more than the highest index.

This is an example of how we can end up with "off by one errors" in programming, which are very common.



````python
i = 25   # assign a variable
s[i]     # this works!
````

````python
s[ len(s)-1 ]  # this works, too!
````

````python
# here's an even easier way:

s[-1]   # negative indexes count from the *right*, not the left!
````

````python
# search in the string -- the "in" operator returns True/False
# the small thing on the left is searched for in the big thing on the right

'j' in s
````

````python
'jkl' in s  # this doesn't mean that j, k, and l are all in s, but that 'jkl' is!
````

````python
'ace' in s
````

````python
# I can get a "slice" of a string, too
# meaning: I can get from a particular index, up to (not including) another index.

s[10:20]   # start at s[10] stop before s[20]
````

````python
s[5:22]    # start at s[5], stop before s[22]
````

````python
s[:15]     # start at the beginnig, stop before 15
````

````python
s[15:]    # start at 15, go through end
````

# However, strings are *immutable*!

You cannot change a string. Ever.

That is: You can assign a new value to a variable referring to a string.

But you cannot change the string itself.

````python
s[0] = '!'   # this is illegal!
````

````python
s = s[15:]    # this is fine: assign s a new value, which happens to be s[15:]

s
````

# Exercise: Pig Latin!

Pig Latin is a children's "secret" language. (It's only secret because children forget that theie parents were once children, too, and probably learned it.)

The rule for translating a word from English into Pig Latin:

- Check the first letter of a word. If it's a vowel (a, e, i, o, or u) then add `way` to the word.
- In all other cases, move the first letter to the end, and then add `ay`.

Some examples:

- `computer` -> `omputercay`
- `elephant` -> `elephantway`
- `desk` -> `eskday`
- `octopus` -> `octopusway`
- `papaya` -> `apayapay`

1. Write a program that asks the user to enter a word. (No capitals, no punctuation, no spaces.)
2. Print the translations of that word into Pig Latin.

````python
word = input('Enter word: ')

if word[0] == 'a' or word[0] == 'e' or word[0] == 'i' or word[0] == 'o' or word[0] == 'u':
    print(word + 'way')
````
