# Loops

Let's say that I want to retrieve each of the elements from a string. 

```python
s = 'abcd'

print(s[0])
print(s[1])
print(s[2])
print(s[3])
```

Unfortunately, this works!

We are repeating ourselves when writing this code -- and that's generally a sign that we're doing something wrong. It's a bad idea to repeat yourself, as the acronym DRY (don't repeat yourself!) says.

Why not? 

1. It takes longer to write.
2. (the big reason) It takes longer to read, understand, and debug the code.
3. (another big reason) If (when!) you have to make changes, you'll then need to make changes in multiple places.
4. it's sometimes faster/more efficient for a programming language, if you don't repeat yourself


# Loops

Loops allow us to tell Python to repeat code multiple times. Python supports two types of loops:

- `for`, in which do something once, for each element of a sequence.
- `while`, in which we do something so long as a condition is `True`

`for` loops are  more common, but we'll see both types in this course.

```python
# re-implement the above code, but using a for loop

s = 'abcd'

for one_character in s:
    print(one_character)
```

# The syntax of `for` loops

1. Start with the keyword `for`.
2. After that, we have the *loop variable*, in this case called `one_character`.
    - The name of the variable is for *YOU* and other people.
    - Python couldn't care less what you call the variable.
    - The name of the variable doesn't affect what the loop does.
    - I could call the variable `one_megabyte`, and it would still work the same way.
3. After the loop variable, we have the word `in`. That's indicating what we're looping over, the *iterable* object.
4. Then we have the iterable
5. Then we have, at the end of the line, `:`. As you saw last week, `:` at the end of a line must be followed by an indented block of one or more lines.
6. The indented block is known as the "loop body," and can contain *any* code, and be as long or short as you want. That is executed once for each element in the iterable.

# How a `for` loop works

Unlike many other languages, Python doesn't make us (writing the `for` loop) keep track of our position in the iterable. Rather, the value itself keeps track of where we are, and provides us with a new value with each iteration.

1. `for` turns to the value at the end of the line, and asks: Are you iterable?
    - If not, the loop exits with an error.
2. Assuming that the value is iterable, `for` asks the value for its next output value.
    - If it's done, then the loop ends, and code continues after the loop body.
3. If there is a value, it's assigned to the loop variable (in this case, `one_character`.)
    - Notice that there isn't any way for `for` to know what we'll get, or even how many values we'll get, when iterating. That is totally up to the iterable value.
4. With the loop variable assigned, the loop body executes.

```python
for one_character in s:
    if one_character != 'c':
        print(one_character)
```

```python
# if you want to count a number of times in Python, you need to use "range"

for counter in range(3):   # here, we aren't iterating over 3, but the value we get back from range(3)
    print('Hooray!')
```

```python
# what is being assigned to counter with each iteration?


for counter in range(3):  
    print(f'[{counter}] Hooray!')
```

When we use `range(n)`, we get `n` iterations. The value we get back from `range` in each iteration is the current count of the loop, starting with 0 and going to up `n-1`.

It never goes up to the value `n` that you state, because it starts with 0.

# Where's the index?

In many other programming languages, a `for` loop doesn't work like in Python. In those languages, you (the coder) need to tell the loop with which number to start, with which number to end, and how much to increment with each iteration.

In those languages, the loop is a mechanism for going from one number to another. You then have to know the index start, end, and what to do with the index number along the way.

Python has a completely different take on this: In Python, you get the characters (or elements) themselves, not their indexes!

This is normally great. But what if you want to show the indexes?

```python
# how can I show each character *and* its index in s?
# one way: manually

s = 'abcd'
index = 0

for one_character in s:
    print(f'{index}: {one_character}')   # print index and character
    index += 1                           # increment the character
```

```python
# option 2: use the enumerate function
# enumerate wraps around an iterable (e.g., a string) and returns *TWO* values, the index + current value

s = 'abcd'

# now have two loop variables!
for index, one_character in enumerate(s):  # now we're iterating over enumerate(s)
    print(f'{index}: {one_character}')   # print index and character
```

```python
# you can kinda sorta do it this way

s = 'abcd'

for index in range(len(s)):
    print(f'{index}: {s[index]}')

# here, though, it's not consider Pythonic -- in the spirit of Python?
# because why use the index to get the characters when (a)the characters are more important,
# (b) Python provides systems to get their values, and (c) we have enumerate
```
