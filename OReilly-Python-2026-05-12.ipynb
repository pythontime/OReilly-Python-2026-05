Why not?

# Loops

Loops allow us to tell Python to repeat code multiple times. Python supports two types of loops:

- `for`, in which do something once, for each element of a sequence.
- `while`, in which we do something so long as a condition is `True`

`for` loops are  more common, but we'll see both types in this course.

```python
# re-implement the above code, but using a for loop

s = 'abcd'

for one_character in s:
    print(one_character)
```

# The syntax of `for` loops

1. Start with the keyword `for`.
2. After that, we have the *loop variable*, in this case called `one_character`.
    - The name of the variable is for *YOU* and other people.
    - Python couldn't care less what you call the variable.
    - The name of the variable doesn't affect what the loop does.
    - I could call the variable `one_megabyte`, and it would still work the same way.
3. After the loop variable, we have the word `in`. That's indicating what we're looping over, the *iterable* object.
4. Then we have the iterable
5. Then we have, at the end of the line, `:`. As you saw last week, `:` at the end of a line must be followed by an indented block of one or more lines.
6. The indented block is known as the "loop body," and can contain *any* code, and be as long or short as you want. That is executed once for each element in the iterable.

# How a `for` loop works

Unlike many other languages, Python doesn't make us (writing the `for` loop) keep track of our position in the iterable. Rather, the value itself keeps track of where we are, and provides us with a new value with each iteration.

1. `for` turns to the value at the end of the line, and asks: Are you iterable?
    - If not, the loop exits with an error.
2. Assuming that the value is iterable, `for` asks the value for its next output value.
    - If it's done, then the loop ends, and code continues after the loop body.
3. If there is a value, it's assigned to the loop variable (in this case, `one_character`.)
    - Notice that there isn't any way for `for` to know what we'll get, or even how many values we'll get, when iterating. That is totally up to the iterable value.
4. With the loop variable assigned, the loop body executes.

```python
for one_character in s:
    if one_character != 'c':
        print(one_character)
```

# Where's the index?

In many other programming languages, a `for` loop doesn't work like in Python. In those languages, you (the coder) need to tell the loop with which number to start, with which number to end, and how much to increment with each iteration.

In those languages, the loop is a mechanism for going from one number to another. You then have to know the index start, end, and what to do with the index number along the way.

Python has a completely different take on this: In Python, you get the characters (or elements) themselves, not their indexes!

This is normally great. But what if you want to show the indexes?

```python
# how can I show each character *and* its index in s?
# one way: manually

s = 'abcd'
index = 0

for one_character in s:
    print(f'{index}: {one_character}')   # print index and character
    index += 1                           # increment the character
```

```python
# option 2: use the enumerate function
# enumerate wraps around an iterable (e.g., a string) and returns *TWO* values, the index + current value

s = 'abcd'

# now have two loop variables!
for index, one_character in enumerate(s):  # now we're iterating over enumerate(s)
    print(f'{index}: {one_character}')   # print index and character
```

```python
# you can kinda sorta do it this way

s = 'abcd'

for index in range(len(s)):
    print(f'{index}: {s[index]}')

# here, though, it's not consider Pythonic -- in the spirit of Python?
# because why use the index to get the characters when (a)the characters are more important,
# (b) Python provides systems to get their values, and (c) we have enumerate
```

As you might know, a decimal number (e.g., 9753) can be described as:

- 9 * 1,000
- 7 * 100
- 5 * 10

Or we can write that using powers of 10:

- 9 * 10**3

1. Ask the user to enter a number
2. Print that number broken down into digits and powers of 10, as you see above
3. You'll want to use a `for` loop, and it'll be super helpful to use `enumerate`. You'll want to use `len` at some point, to get the length of the full string you got from the user.

```python
number = input('Enter a number: ').strip()   # remember, number is a string!

for index, one_digit in enumerate(number):  # use enumerate to get each digit + its position (index)
    power = len(number) - index - 1         # use len(number) and index to calculate the power
    print(f'{one_digit} * 10 ** {power}')   # print it using an f-string
```

# Controlling our loops

So far, we've seen that we can use a `for` loop to iterate over a number of elements of a string (or a `range`). And the loop runs one iteration on each element of the iterable, until we get to the end.

But if we don't need to get to the end? What if we don't want to?

For that, Python provides two statements:

- `break` stops the loop right away, and program execution continues after the loop body ends.
- `continue` stops the current iteration right away, continuing with the next iteration (if there is one)

When do we use these?

- `break` is typically used when we are in a loop, and we realize there's no reason to continue. We can just stop the loop and go one.
- `continue` is typically used when the current iteration isn't going to help or solve our problem, and we want to move onto the next one.

Both are almost always used inside of `if` statements. If such-and-such condition is `True`, then we `break` or `continue`.

```python
s = 'abcde'
look_for = 'd'

print('Starting')
for one_character in s:
    if one_character == look_for:
        break   # exit the loop, continuing afterwards
    print(one_character)
print('Ending')
```

```python
s = 'abcde'
look_for = 'd'

print('Starting')
for one_character in s:
    if one_character == look_for:
        continue   # exit the current iteration, continuing with the next one
    print(one_character)
print('Ending')
```

3. Go through each character in the string:
    - If we encounter `.`, stop everything and print `total`.
    - If we encounter a non-digit (other than `.`), then scold the user and go on.
    - Turn the digit into an integer, and add to `total`.
4. Print `total`

The idea is that we'll add the digits in a string, ignoring non-digits. But if there is a `.`, then we stop prematurely.

Example:

    Enter characters to sum: 12a3b4.56

    adding 1, total is 1
    adding 2, total is 3
    ignoring a, which is not numeric
    adding 3, total is 6
    ignoring b, which is not numeric
    adding 4, total is 10
    stopping, thanks to .

    total is 10

```python
total = 0

text = input('Enter text: ').strip()

for one_character in text:

    if one_character == '.':
        print(f'Got . -- stopping now')
        break

    if one_character.isdigit():
        n = int(one_character)
        total += n
        print(f'added {n}; total is {total}')
    else:
        print(f'{one_character} is not numeric; ignoring')

print(total)
```

# `while` loops

A `for` loop is great when you know how many times you want to iterate, or when you have a sequence of values, and want to perform the same operation on all of them.

But sometimes, you don't know how many iterations you'll need. In such a case, you can't use `for`. Instead, you use `while`.

You can think of a `while` loop as an `if` statement that repeats until the condition is `False`. You put a condition after `while`. If it's `True`, then the block executes. If it's `False`, then we skip over the entire loop body.

The loop body runs, and then we go back to the `while`, and check the condition again.

For this reason, the condition should include a variable or expression whose value can change! Using an unchanging value means that you might well get into an infinite loop.

```python
x = 5

while x > 0:
    print(x)
    x -= 1   # same as: x = x - 1
```

```python
# sometimes, we might want an infinite loop -- just make sure there's an escape hatch

while True:   
    name = input('Enter your name: ').strip()

    if name == '':  # empty string?
        break

    print(f'Hello, {name}!')
```

    ignoring asdfasdf, which is not a number
    Adding 40, total is 100

```python
total = 0
text = input('Enter numbers: ').strip()

for one_item in text.split():  # splitting on whitespace, gives us a list
    if one_item.isdigit():
        total += int(one_item)
        print(f'Added {one_item}, total is now {total}')
    else:
        print(f'{one_item} is not numeric; ignoring.')

print(f'total = {total}')       
```

# Tuples

We've seen two "sequence" types in Python already:

- Strings, which contain characters and are immutable
- Lists, which contain *anything* and are mutable

Tuples are the third member of the "sequence" family. They combine these:

- Tuples can contain anything (like lists)
- Tuples are *immutable*, like strings

It's very common to think of tuples are "locked lists," or immutable lists... but that's not the way you're supposed to think about them. Rather, you're supposed to think that if you have a bunch of values of the same type, they should be in a list. If you have a bunch of values of different types, they should be in a tuple. 

Why would you use a tuple?

- Tuples are Python's "records" or "structs." The idea is that if you have a few values, not of the same type, the tuple is a place to store it.
- Python uses tuples a lot behind the scenes, so even if you don't, Python will.
