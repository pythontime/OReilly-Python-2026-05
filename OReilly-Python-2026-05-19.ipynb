# What are dicts?

A dictionary in Python is a very old data structure, sometimes known by other names:

- Hash tables
- Hash maps
- Maps
- Hashes
- Key-value stores
- Name-value stores
- Associative arrays

This data structure is basically a two-column table. The column on the left is known as the "keys" and the column on the right is known as the "values."

Another way to think about dicts is that they're like lists, except that the dict doesn't dictate what the indexes will be. In a list, the indexes are 0, 1, 2, etc. But in a dict, you can determine what the indexes are; they're known as "keys," and can be any immutable value in Python.

# Rules for dict keys and values

## Keys
- Any immutable value in Python (i.e., numbers + strings + tuples are OK, but lists and dicts are not)
- In a given dict, the keys must be unique
- Every key has a value, and every value has a key

## Values
- There are no limitations; any value of any type can be a value in the dict. They can even repeat!

# Avoiding errors

Everything in a dict is done via the keys. You get the values via the keys. You cannot get a key via its value.

The keys in a dict are primary. For that reason, you can check whether a key is in a dict with the `in` operator. This operator also works on strings, lists, and tuples. But in those data structures, we check if an element is there. In the case of dicts, we're asking if the *key* is there.

```python
'a' in d   # is 'a' a key in d?
```

```python
'x' in d  # is 'x' a key in d?
```

```python
10 in d   # is 10 a key in d?  No, but it is a value
```

```python
menu = {'sandwich':10, 'tea':5, 'apple':1, 'cake':3}
total = 0

while True:  # infinite loop
    order = input('Order: ').strip()

    if order == '':    # empty order?
        break

    if order in menu:         # is the user's order string a key in the menu dict?
        price = menu[order]   # get the value for a key with d[key]
        total += price        # add the price to total
        print(f'{order} costs {price}; total is now {total}')
    else:
        print(f'Sorry, we are out of {order} today!')

print(f'Total is {total}')        
```

# Are dicts mutable?

We've seen that strings are immutable -- once you have defined a string value, you cannot change it. You *can* assign new values to a variable that referred to a string, but the string itself cannot be changed.

By contrast, we've seen that lists are *mutable* -- they can be changed. You don't need to assign or re-assign a variable if you've changed a list in order to see a difference.

There are three parts to mutability:

1. You can update an existing element
2. You can add a new element
3. You can remove an existing element

```python
mylist = [10, 20, 30, 40, 50]

mylist[2] = 999   # updating an element
mylist
```

```python
mylist.append(12345)  # added a new element, adding to the length of the list
mylist
```

```python
mylist.pop()   # removes + returns the final element, shortening the list
mylist
```

```python
# what about dicts? They are mutable!

# how do we update/modify/replace an existing element? Just assign to it, using the key
d
```

```python
d['a'] = 999  # replaced the value for this key
d
```

```python
# can we add a new key-value pair?
# yes! And the syntax is... exactly the same as for replacing
# if the key exists, we replace. If not, we create

d['x'] = 98765   # added a new key-value pair
d
```

```python
# can we remove an existing key-value pair?
# yes, using the dict.pop method -- which is similar to the list.pop method

d.pop('x')   # this removes the key-value pair for x, and returns the value
```

# Accumulating with dicts

It's very common to create a dict with keys and empty values (e.g., 0) when we start. Then, over the course of the program, we add to the value, and accumulate.

Remember you can say

    d['a'] += 1

# Exercise: Vowels, digits, and others (dict edition)

1. Define a dict, `counts`, with three keys -- `vowels`, `digits`, and `others`. All three values should be 0.
2. Ask the user to enter a string.
3. Go through the string, one character at a time:
    - If it's a vowel (a, e, i, o, u) then add 1 to `counts['vowels']`
    - If it's a digit 0-9, then add 1 to `counts['digits']`
    - Otherwise, add 1 to `counts['others']`.
4. Print `counts`.

```python
counts = {'vowels':0,
          'digits':0,
          'others':0}

text = input('Enter text: ').strip()

for one_character in text:
    if one_character in 'aeiou':    # found a vowel?
        counts['vowels'] += 1       # ... add 1 to the existing value of counts['vowels'], and assign to the value
    elif one_character.isdigit():   # found a digit?
        counts['digits'] += 1       # ... add 1 to counts['digits']
    else:
        counts['others'] += 1

print(counts)
```

# Accumulating the unknown

What if we want to have a dict that counts data, but we don't know what we want to count?

That is: We don't know what inputs we'll get. And we either cannot or don't want to pre-populate the dict.

For example: I want to count how many times each character appears in a string.

```python
counts = {}   # empty dict

text = input('Enter text: ').strip()

for one_character in text:
    counts[one_character] += 1    # this will fail -- one_character isn't a key in counts!
```

```python
# we'll have to check whether the key already exists.
# - if so, then add to the value, not a problem
# - if not, then we add the key-value pair

counts = {}   # empty dict

text = input('Enter text: ').strip()

for one_character in text:
    if one_character in counts:   # have we seen this key before? 
        counts[one_character] += 1   
    else:
        counts[one_character] = 1   # initialize this key-value pair

print(counts)        
```

Why not create a dict whose keys are *all* of the potential characters?

That'll just waste time and memory.

The goal of this exercise is to get names of cities and daily rainfall for that city from the user. We don't know in advance what cities the user will enter, and so our dict cannot be pre-populated.

1. Create an empty dict, `rainfall`.
2. Ask the user for the name of the city for which we're reporting.
    - If the user enters an empty string, stop asking (i.e., `break` from the loop)
3. If we did get a city name, ask a second question, i.e., how many mm of rain fell?
    - We can assume the user will give us digits
4. Take that city name and rainfall amount:
    - If we've seen the city before, add to the existing rainfall
    - If it's a new city, then add a new key-value pair to our dict, where the key is the city and the value is `mm_rain`.

Example:

    City: a
    Rain: 5
    City: b
    Rain: 4

    Rain: 3
    City: [ENTER]
    {'a':8, 'b':4}

```python
rainfall = {}   # empty dict -- we know keys will be strings (city names) and values will be ints (mm rain)

while True:     # I don't know how many reports I'll get
    city_name = input('City: ').strip()

    if city_name == '':    # no city? no continued looping!
        break

    mm_rain = input('Rain: ').strip()
    mm_rain = int(mm_rain)  # dangerous, but I'll allow it here

    if city_name in rainfall:            # have we seen city_name in rainfall before?
        rainfall[city_name] += mm_rain   # add to the existing value
    else:
        rainfall[city_name] = mm_rain    # add the key-value pair to our dict

print(rainfall)      
    
```

# Iteration and dicts

We've seen that Python encourages us to use `for` loops all over, and that data structures all support `for` loops:

- Iterate over a string, and get each character
- Iterate over a list or tuple, and get each element

What happens if we iterate over a dict? Will it work? What will we get?

```python
d = {'a':10, 'b':20, 'c':30}

for one_thing in d:
    print(one_thing)
```

```python
# what if I want to print a dict's keys and values?


for one_key in d:
    print(f'{one_key} : {d[one_key]}')
```

There's nothing technically wrong with this! But... wouldn't it be nice if we could get both the keys and values?

Yes, there is a method, `dict.items`, which is meant for precisely this purpose. With each iteration, we get a 2-element tuple, `(key, value)`.

```python
for one_pair in d.items():
    print(one_pair)
```

```python
# use tuple unpacking

for one_pair in d.items():
    key, value = one_pair
    print(f'{key}: {value}')
```

```python
# unpack in the for loop itself
# this is my favorite way to iterate over a dict

for key, value in d.items():
    print(f'{key}: {value}')
```

```python
# you'll sometimes see people iterating over dict.keys()
# if you just want the keys, you can do that!

for one_key in d.keys():   # don't do this -- it's better to just iterate over d itself
    print(one_key)
```

```python
# same thing for searching

'a' in d.keys()   # this will work... but why do it?
```

```python
'a' in d    # shorter, faster, why use the method?
```

# Ordering in dicts

When you iterate over a dict, you get the key-value pairs in the order that the keys were added. Normally, you don't need to care about the order of a dict, and for many years, you *couldn't* depend on the order.

```python
d = {'c':30, 'b':20, 'a':10}

for key, value in d.items():
    print(f'{key}: {value}')
```

```python
d.pop('b')  # removes the key-value pair with 'b' as a key
```

```python
d['b'] = 999  # add a new key-value pair 
```

```python
for key, value in d.items():
    print(f'{key}: {value}')
```
