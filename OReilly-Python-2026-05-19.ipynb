# Dictionaries ("dicts")

We've seen over the previous two weeks that Python uses lots of data structures

Each value in Python has a different type, representing a different data structure.

So far, we've used:

- Numbers (`int` and `float`)
- Text (`str`)
- Sequences (`list` and `tuple`)

Dictionaries are, without a doubt, the most important data structure in Python. They are high speed, flexible, easy to work with, and generally help us solve a lot of different problems.
