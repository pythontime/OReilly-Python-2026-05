# Some useful methods

- Strings
    - `str.strip` -- returns a new string, without surrounding whitespace
    - `str.split` -- returns a list of strings, broken up on a delimiter/separator character
    - `str.lower` -- returns a new string, lowercase from the original
    - `str.isdigit` -- returns `True` if the string is non-empty and contains only 0-9
    - `str.join` -- returns a string, based on a "glue" string and a list of strings, e.g., `'*'.join(['a', 'b', 'c'])`
- Lists
    - `list.append` -- adds a new value to the end of the list
    - `list.pop` -- removes and returns the final element from a list

# Dictionaries ("dicts")

We've seen over the previous two weeks that Python uses lots of data structures

Each value in Python has a different type, representing a different data structure.

So far, we've used:

- Numbers (`int` and `float`)
- Text (`str`)
- Sequences (`list` and `tuple`)

Dictionaries are, without a doubt, the most important data structure in Python. They are high speed, flexible, easy to work with, and generally help us solve a lot of different problems.

# What are dicts?

A dictionary in Python is a very old data structure, sometimes known by other names:

- Hash tables
- Hash maps
- Maps
- Hashes
- Key-value stores
- Name-value stores
- Associative arrays

This data structure is basically a two-column table. The column on the left is known as the "keys" and the column on the right is known as the "values."

Another way to think about dicts is that they're like lists, except that the dict doesn't dictate what the indexes will be. In a list, the indexes are 0, 1, 2, etc. But in a dict, you can determine what the indexes are; they're known as "keys," and can be any immutable value in Python.

# How to create a dict?

- Use `{}` to create one
- Each key-value pair has a `:` between the key and the value
- The pairs are separated with `,` like in a list or tuple

```python
d = {'a':10, 'b':20, 'c':30}  # this is a dict with 3 key-value pairs
```

```python
type(d)  # what kind of value is this?
```

```python
len(d)  # what is the length of d -- how many pairs are in it?
```

# Rules for dict keys and values

## Keys
- Any immutable value in Python (i.e., numbers + strings + tuples are OK, but lists and dicts are not)
- In a given dict, the keys must be unique
- Every key has a value, and every value has a key

## Values
- There are no limitations; any value of any type can be a value in the dict. They can even repeat!

```python
d
```

```python
d['a']    # notice that I use [] to retrieve the value based on the key
```

```python
k = 'b'   # define a variable
d[k]      # retrieve based on the key, stored in k
```

```python
d['something else']   # this gives us a KeyError exception, meaning that the key didn't exist
```

When you retrieve based on the key, what you put in square brackets must match the key PRECISELY. Case and whitespace are important!

```python
d['a ']   # this will fail!
```

```python
d['A']   # this will also fail!
```

# Can you use a variable as a key?

Variables aren't values themselves. They refer to values. You can use a variable as a key, but the value to which it refers will be used.

```python
k1 = 'a'

d[k1]  # this and d['a'] are seen as precisely the same by Python
```

```python
d['a']
```
