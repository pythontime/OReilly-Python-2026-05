# What are modules?

We've talked several times about the "DRY rule," meaning, "Don't repeat yourself." 

- If the same line is repeated several times in a row, we can replace it with a loop
- If the same functionality is repeated several times in the same program, we can write and use a function.
- If the same functionality is repeated several times in *multiple* programs? There, we use a *library*.

In Python, we use the term *module* instead of library. A module actually does two different things:

- It provides libraries -- meaning that we can write code once, put it into a module, and then use it in multiple programs
- It also provides *namespacing* -- this avoids the potential for "namespace collisions."

If your program uses a variable `x`, or defines a function named `hello`, and you import a module that has *also* defined a variable `x` or a function `hello`, which gets priority? This can be a real pain to identify and debug.

In Python, every module is its own "namespace," meaning its own container for names. If you use `x` from a module, it won't be known as `x`, but rather `themodule.x`. You can think of `themodule` in this example as a module name, but also as a namespace, sort like a last name for variables (for functions, too).

If you import modules `A` and `B` into your program, you don't want to have to think about whether `A` and `B` have overlapping, and thus colliding, names!

The bottom line is: If you want to use someone else's functionality, then they can (should) provide it in the form of a module.

If you have implemented some functionality, and want others to use it, you can provide it as a module to them.

In many ways, using modules are one of Python's easiest to use functions. Writing modules is more complex, and distributing them to other people is even more complex.

# Let's use a module!

If I want to use a module in Python, I use the `import` statement. This loads a module's contents from disk (if needed), creates a new in-memory module value in Python (reflecting the file's contents), and then assigns that module value to a variable.

````python
import random
````

# A few things to notice here

1. `import` is a statement. It is also a reserved word, meaning you cannot redefine it. It is not a function! You don't use `()` to invoke it.
2. The name that we put after `import` serves two purposes. It both is the name of the new variable that will refer to the module value that's created. It's also the name of the file that will be loaded into memory. Because we're importing `random`, Python will look for the file `random.py`. (Yes, the `.py` is mandatory.)
3. Many programming languages work differently when it comes to loading modules. In those languages, you give a string -- a filename! -- to the `import` equivalent, and the language loads that file. Not so in Python! You cannot give it a filename. Instead, you give it a module name, and it figures out the filename from that.

# Exercise: Guessing game

1. `import random`, to load the `random` module into memory.
2. Use the `random.randint` function to choose a secret number between 0-100.
3. Ask the user, repeatedly, to guess the secret number.
    - If they get it right, exit from the program, printing the number of guesses they needed.
    - If it's too high, say "too high!"
    - If it's too low, say "too low!"
4. Print the number of guesses.

Example:

    Guess: 50
    Too high!
    Guess: 25
    Too low!
    Guess: 37
    You got it in 3 guesses.


````python
import random

secret = random.randint(0, 100) 
guess_count = 0

while True:
    guess_text = input('Guess: ').strip()

    if not guess_text.isdigit():
        print(f'{guess_text} is not numeric; try again')
        continue

    guess = int(guess_text)
    guess_count += 1
    
    if guess == secret:
        break
    elif guess < secret:
        print('Too low!')
    else:
        print('Too high!')

print(f'You got it in {guess_count} tries.')
````

# What if I just want to run `randint`?

It's kind of annoying to always have to write `random.randint`. We're all friends here; why not just invoke `randint`?



````python
randint(0, 100)
````

Names in a module are really only accessible via the module. You cannot `import random` and then ignore the `random` module name. If you've imported `random`, then every name is `random.SOMETHING`.

There is, however, a way around this! We can use the `for .. import` statement.

Instead of saying

    import random
    n = random.randint(0, 100)

we can say

    from random import randint
    n = randint(0, 100)    # look, no "random." before "randint"!

# Things to know about `from .. import`

1. The entire module, from start to finish, is loaded into memory.
2. For this reason, we can say that it doesn't use less memory.
3. Python assigns global variables to the same module's names that we've specified.
4. Why use `from .. import`, then? Because it makes it easier to write code when you just need one name, not two.
5. When you use `from .. import`, then the module is loaded into memory *BUT* the module variable isn't defined. So you cannot access other functions/variables in the module until you import them, too.

There are places that say: Using `from .. import` is a bad idea, because it removes the context of the module name. If you invoke `randint` in your code, then do you really know which module it comes from?

# Exercise: Another guessing game!

1. Define a list of strings, where each string represents a month of the year. (There will be 12 strings in the list.)
2. Use `random.choice`, a function in `random` to select one of the months. However, you don't want to invoke it as `random.choice`. Rather, you want to invoke it as `ch` for short. You'll invoke `ch` on your list of strings to get a random one.
3. Ask the user, repeatedly, to guess the month that was selected.
4. When they answer correctly, indicate how many guesses they needed.

````python
from random import choice as ch

months = 'Jan Feb Mar Apr May Jun Jul Aug Sep Oct Nov Dec'.split()
random_month = ch(months)
guess_counter = 0

print(f'The secret month is {random_month}')

while True:
    guess = input('Guess a month: ').strip().capitalize()[:3]

    guess_counter += 1
    if guess == random_month:
        break

print(f'You got it in {guess_counter} tries!')       
````

````python
# Marcos

from random import choice as ch

months = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

secret = ch(months)
guesses = 0

while True:
    guess = input("Guess the month: ").strip()
    if not guess:
        continue
    guesses += 1

    if guess.casefold() == secret.casefold():
        print(f"Correct! It took you {guesses} guess(es).")
        break
    else:
        print("Nope,don't be lazy try again.")
````
