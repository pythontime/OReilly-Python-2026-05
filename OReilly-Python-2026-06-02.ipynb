# What are modules?

We've talked several times about the "DRY rule," meaning, "Don't repeat yourself." 

- If the same line is repeated several times in a row, we can replace it with a loop
- If the same functionality is repeated several times in the same program, we can write and use a function.
- If the same functionality is repeated several times in *multiple* programs? There, we use a *library*.

In Python, we use the term *module* instead of library. A module actually does two different things:

- It provides libraries -- meaning that we can write code once, put it into a module, and then use it in multiple programs
- It also provides *namespacing* -- this avoids the potential for "namespace collisions."

If your program uses a variable `x`, or defines a function named `hello`, and you import a module that has *also* defined a variable `x` or a function `hello`, which gets priority? This can be a real pain to identify and debug.

In Python, every module is its own "namespace," meaning its own container for names. If you use `x` from a module, it won't be known as `x`, but rather `themodule.x`. You can think of `themodule` in this example as a module name, but also as a namespace, sort like a last name for variables (for functions, too).

If you import modules `A` and `B` into your program, you don't want to have to think about whether `A` and `B` have overlapping, and thus colliding, names!

The bottom line is: If you want to use someone else's functionality, then they can (should) provide it in the form of a module.

If you have implemented some functionality, and want others to use it, you can provide it as a module to them.

In many ways, using modules are one of Python's easiest to use functions. Writing modules is more complex, and distributing them to other people is even more complex.
