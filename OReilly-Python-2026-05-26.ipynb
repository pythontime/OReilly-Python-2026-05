# What are functions?

The data in a programming language, such as Python, are our *nouns*. These are the values that we deal with on a day-to-day basis. We've seen a variety of types of data that we can work with -- strings, lists, tuples, dicts, etc.

If you want to do something in a human language, then you need *verbs*. The same thing is true in Python. So far, we've learned about a lot of different. Some of the functions we've seen are:

- `print`
- `input`
- `len`
- `sum`
- `type`

We don't need these functions! We could implement everything ourselves. But why re-invent the wheel? If functionality is going to be repeated very often in our program, or across our programs, then it makes sense to define it once and use it many times.

This is the DRY (don't repeat yourself) rule! Python provides us with a bunch of functions (and methods) that allows us to avoid defining them each time we want to write a program. We can concentrate on what's important to us, and what's different about our program.

We previously discussed the DRY rule when we talked about loops.

But there's another reason for using functions, too:

# Abstraction

The idea behind abstraction is that you cannot think about all of the little details in all of the little parts, all of the time. Instead, we can wrap up a bunch of complex functionality, put it in a box, label the box, and then treat that box as a small building block.

Abstraction allows us to wrap our heads around complex mechanisms, and replace the complexity with one small item. It also allows us to communicate with other people about what we want to do.

What we can do with functions is take a number of steps, and wrap them into a single step, one function's name. We can then invoke that function, and get a whole bunch of things to happen.

Defining a function in Python defines a new verb in the language. But we don't *need* to define functions to have all of the functionality we want. That said, without functions, it would be very difficult for us to reason about our programs, and also difficult for us to communicate about them with other people.

Defining functions is one of the most common actions that a programmer makes. 

# Let's define a function!

To define a function:

1. We use the keyword `def`, which means: I want to define a new function.
2. Immediately after `def`, we give the function a name. This is actually a variable assignment; the function's name is a variable!
3. We then put `()` after the function name. Right now, these will remain empty, but pretty soon, we'll put parameter names between the `()`.
4. At the end of the line, we use `:`
5. Then we have one or more lines in an indented block. This is known as the "function body." You can put *ANY CODE AT ALL* inside of the function body. You can use `if`, `for`, `while`, `input`, or `print`.

The function body doesn't execute when you define the function. When you want to execute the function's body, you invoke the function by naming it and putting `()` after the name. If you don't put the `()` there, then the function won't run.

```python
for i in range(3):
    hello()
```

```python
def hello():
    name = input('Enter your name: ').strip()
    print(f'Hello, {name}!')
```

When you invoke a function, you can pass one or more arguments to it:

- `len('abcd')` -- passed one argument of type `str`

Most programmers confuse two terms:

- arguments are the values that we pass when we invoke a function
- parameters are the variables to which arguments are assigned

When you define a function, you can define it with one or more parameters in the `()` after the function name. That determines how many parameters the function has, which in turn determines how many arguments the user must pass.

```python
# now we can invoke "hello" with an argument, and avoid having to type, to give "input" a value

def hello(name):
    print(f'Hello, {name}!')  

hello('world')     # the string argument 'hello' was assigned to the parameter name -- which then can be used in the body 
```

# Dynamic typing

In Python, values have types. So you can have a string value, or a list value, or a dict value.

However, variables don't have types. Any variable can refer to any value, and can be reassigned to different value (of a different type) at any time.

```python
x = 'abcd'   
type(x)  # what kind of value is x referring to?
```

```python
x = 12345    # we've reassigned x to refer to an integer
type(x)  
```

In a *statically typed* language, the variables have types, too! When you define a variable, you have to "declare* what type of value it will contain. If you try to assign the wrong kind of value, you'll get an error. You'll probably get an error before the program even runs, since the language will monitor this kind of mistake, and will stop you before the program even runs.

People who come from a background of static typing *love* that sort of thing. Moreover, it basically avoids a whole class of bugs.

Python doesn't do this. Python is dynamically typed. You can assign any value to any variable, and the language won't complain. However, it will complain if the value doesn't support the method you want to invoke, or the operator you want to use.

```python
x = '12345'
y = '67890'

x.isdigit() and y.isdigit()
```

```python
# this won't work, because neither x nor y refers to a string
x = 12345
y = 67890

x.isdigit() and y.isdigit()
```

When we define a function, we don't have a good way to indicate, "This parameter must always get that type of value." 

You can't say that `first` and `second` must both be integers, and `op` must be a string. 

You have to document your function, to indicate to whoever might call it what types of values to pass. But you're basically trusting the caller to read the documentation and to follow it.

What's the advantage of dynamic typing? It means that we can write one function, one time, that handles many different types of values. `len`, for example, can handle strings, lists, tuples, and dicts. 

Python is now used by very large teams at very large companies, on projects that are worth very large amounts of money. 

Over the last number of years, Python has introduced "type hints," which allow us to tag parameters and variables with a particular type.

```python
def add(x:int, y:int):   # these are "type hints," and you can set them... but Python (the language) ignores them.
    print(x + y)
```

# Return values

So far, our functions have printed something on the screen. But they haven't *returned* a value to the caller.

# How can we return a value?

Our function can return a value whenever it wants with the `return` statement.

`return` does two things, actually:

1. It returns the value we specify to the caller
2. It stops the function body from executing, and exits from it

A function can have multiple `return` statements; the first one to execute is also the only one to execute.

```python
def myfunc():
    return 10

x = myfunc()
print(x)   # what's x's value?
```

# Keyword arguments

So far, when we've invoked functions, we have passed *positional arguments*. This means that each argument is assigned to the parameter in the same position.

If I have parameters a, b, and c, then calling a function with (10, 20, 30) assigns to them in parallel.

It turns out that there is another way to pass arguments, and this other way assigns them differently to parameters.

Those are *keyword arguments*.

Keyword arguments all have the form of `name=value`. When you invoke the function, you don't just pass the value; you also specify the parameter that should accept that value.

The assignment to parameters is done via the keyword, not via the position.

You can mix and match positional arguments and keyword arguments. However, all positional arguments must come before all keyword arguments.

```python
# parameters: a   b   c 
# arguments:  10  20  30

def myfunc(a, b, c):
    return f'a = {a}, b = {b}, c = {c}'

myfunc(10, 20, 30)      # how do I know these are positional? No name=value arguments.
```

```python
# parameters: a   b   c 
# arguments: 10   20  30 

def myfunc(a, b, c):
    return f'a = {a}, b = {b}, c = {c}'

myfunc(a=10, b=20, c=30)      # these are all keyword arguments!
```

```python
# you can pass keyword arguments in any order!

# parameters: a   b   c 
# arguments: 

def myfunc(a, b, c):
    return f'a = {a}, b = {b}, c = {c}'

myfunc(b=10, c=20, a=30)      # these are all keyword arguments!
```

```python
# can I mix positional + keyword?
# yes, so long as ALL positional come before ALL keyword

myfunc(10, b=20, c=30)  # not a problem!
```

```python
myfunc(a=10, 20, 30)  # not OK!
```

# Default argument values

It can be annoying to discover that if you want to invoke a function, it requires a lot of arguments. Or maybe just more arguments than you would like.

This is especially true if a function is normally used in a particular way; why should you have to pass the same argument to it every time?

For this, Python provides *default argument values*. That is: When you define the function, you can indicate what the default argument should be for a parameter. If the parameter isn't assigned an argument, the default is retrieved and used.

```python
def add(x, y):
    return x + y
```

# Rules for default argument values

1. When a parameter has a default argument value, it's effectively an optional parameter. If you don't specify it in the function call, then it gets the default value.
2. When you define the function, all mandatory parameters (i.e., without defaults) must come before all optional parameters (i.e., with defaults).
3. You can, in theory, have any value as a default argument but in practice you should only use immutable values. In other words, don't use lists and dicts as default argument values.
4. If a function has a number of parameters with defaults, and you invoke the function with keyword arguments, you can "skip over" one or more parameters, letting them use their defaults.

```python
# try the wrong way...

def add(x=10, y):     
    return x + y
```

- Add another parameter, `lowercase`, which is a `True`/`False` value, defaulting to `False`. If this is `True`, then make the comparison case-insensitive by running `lower` on the string before looking for vowels (or whatever).



```python
def count_characters(text, to_count='aeiou', lowercase=False):
    total = 0

    if lowercase:
        text = text.lower()
        to_count = to_count.lower()
    
    for one_character in text:
        if one_character in to_count:
            total += 1
    return total

count_characters('Hello To Everyone Out There')  # how many vowels?  
```

```python
count_characters('Hello To Everyone Out There', lowercase=True)
```

```python
count_characters('Hello To Everyone Out There', to_count = 'abcde')
```
